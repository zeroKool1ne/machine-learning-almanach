```python
# Musterlösung:
from sklearn.datasets import load_wine
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

wine = load_wine()
X, y = wine.data, wine.target

modelle = {
    'Tree (kein Limit)': DecisionTreeClassifier(random_state=42),
    'Tree (depth=3)': DecisionTreeClassifier(max_depth=3, random_state=42),
    'Random Forest (10)': RandomForestClassifier(n_estimators=10, random_state=42),
    'Random Forest (100)': RandomForestClassifier(n_estimators=100, random_state=42),
}

for name, m in modelle.items():
    s = cross_val_score(m, X, y, cv=5)
    print(f"{name:<25} {s.mean():.4f} ± {s.std():.4f}")
```
