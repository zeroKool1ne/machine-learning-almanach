## Aufgabe 2 — Discount-Faktor verstehen

Ein Agent kann eine Belohnung von **+10** erhalten — entweder:
- **Sofort** (Schritt 0)
- **In 5 Schritten** (Schritt 5)
- **In 10 Schritten** (Schritt 10)

Berechne den abgezinsten Wert für γ = 0.9 und γ = 0.5.

**Formel:** `abgezinster Wert = γ^t × Belohnung`

````python
# Deine Lösung:
belohnung = 10
schritte  = [0, 5, 10]

# Berechne und plotte den abgezinsten Wert für gamma=0.9 und gamma=0.5
# Was bedeutet das für den Agenten?
````

<details>
<summary>Musterlösung</summary>

```python
for gamma in [0.9, 0.5]:
    print(f"\nγ = {gamma}:")
    for t in schritte:
        wert = (gamma ** t) * belohnung
        print(f"  Schritt {t:2d}: {wert:.3f}")

# γ=0.9: Schritt 0=10.0, Schritt 5=5.9, Schritt 10=3.49
# γ=0.5: Schritt 0=10.0, Schritt 5=0.31, Schritt 10=0.01

# Mit γ=0.5 ist eine Belohnung in 10 Schritten fast wertlos!
# → Agent mit kleinem γ denkt sehr kurzfristig
```

## Aufgabe 3 — ε-Greedy Strategie

Implementiere eine ε-Greedy-Funktion und beobachte wie sich ε auf die Exploration auswirkt.

````python
# Q-Werte für 4 Aktionen in einem Zustand
q_values = [2.0, 5.0, 1.0, 3.5]
actionen = ['Links', 'Rechts', 'Hoch', 'Runter']

# Aufgabe:
# 1. Implementiere epsilon_greedy(q_values, epsilon) → gibt Index der gewählten Aktion zurück
# 2. Simuliere 1000 Schritte für epsilon in [0.0, 0.1, 0.3, 0.9]
# 3. Plotte als Balkendiagramm: Wie oft wurde jede Aktion gewählt?
# 4. Welche Aktion sollte bei epsilon=0.0 immer gewählt werden?

# Deine Lösung:
````

<details>
<summary>Musterlösung</summary>

```python
def epsilon_greedy(q_values, epsilon):
    if random.random() < epsilon:
        return random.randint(0, len(q_values)-1)
    return int(np.argmax(q_values))

fig, axes = plt.subplots(1, 4, figsize=(14, 4))

for ax, eps in zip(axes, [0.0, 0.1, 0.3, 0.9]):
    counts = [0] * 4
    for _ in range(1000):
        a = epsilon_greedy(q_values, eps)
        counts[a] += 1
    ax.bar(actionen, counts)
    ax.set_title(f'ε = {eps}')
    ax.set_ylabel('Anzahl Auswahl')

plt.tight_layout()
plt.show()

# Bei ε=0.0: immer 'Rechts' (Index 1, Q=5.0 ist höchster Wert)
# Bei ε=0.9: fast gleichmäßige Verteilung → viel Exploration
```
</details>

## Aufgabe 4 — Q-Tabelle manuell updaten

Gegeben:
- `Q(s, a) = 2.0` (aktueller Q-Wert)
- `reward = 5` (erhaltene Belohnung)
- `max Q(s', a') = 3.0` (bester Q-Wert im nächsten Zustand)
- `α = 0.1`, `γ = 0.9`

**Formel:** `Q(s,a) ← Q(s,a) + α × [r + γ × max Q(s',a') - Q(s,a)]`

Berechne den neuen Q-Wert.

````python
# Deine Lösung:
q_alt    = 2.0
reward   = 5
q_next   = 3.0
alpha    = 0.1
gamma    = 0.9

# q_neu = ?
# print(f"Neuer Q-Wert: {q_neu:.4f}")
````

<details>
<summary>Musterlösung</summary>

```python
td_target = reward + gamma * q_next    # 5 + 0.9 * 3.0 = 7.7
td_error  = td_target - q_alt          # 7.7 - 2.0 = 5.7
q_neu     = q_alt + alpha * td_error   # 2.0 + 0.1 * 5.7 = 2.57

print(f"TD-Target: {td_target}")
print(f"TD-Error:  {td_error}")
print(f"Neuer Q-Wert: {q_neu:.4f}")
# → 2.57  (etwas höher als vorher, da Erfahrung positiv war)
```
